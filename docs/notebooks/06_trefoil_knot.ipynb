# Trefoil Knot

Fits a one-dimensional Fourier series to a noisy sample of a trefoil knot embedded in $\mathbb{R}^{3}$.


```python
from dreimac import CircularCoords
from dreimac import GeometryExamples, PlotUtils
from persim import plot_diagrams
import matplotlib.pyplot as plt
import torofit as TF
```

```python
X = GeometryExamples.trefoil(n_samples = 2500, horizontal_width=10)
fig = plt.figure(figsize=(10,6))
ax1 = fig.add_subplot(121)
ax1.scatter(X[:,0],X[:,1], s=1)
ax1.set_title("Trefoil 2D projection") ; ax1.set_aspect("equal") ; ax1.axis("off")
ax2 = fig.add_subplot(122,projection='3d')
ax2.scatter(X[:,0],X[:,1],X[:,2], alpha=0.5, s=1)
ax2.set_title("Trefoil in 3D") ;  _ = PlotUtils.set_axes_equal(ax2)
```

```python
cc = CircularCoords(X, 300, prime=3)
plot_diagrams(cc._dgms)
```

```python
circular_theta = cc.get_coordinates(perc=0.1,cocycle_idx=0)
```

```python
K = 5; lam=1.0

V=TF.fit(X, circular_theta, K, lam)
res = plot_function_via_coeff_matrix(V, K, d=1,samples=1000)
```
